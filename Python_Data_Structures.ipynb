```python
##String Concatenation

greeting="Hello"
name=input("Enter your name:  ")
result=greeting + name
print(result)
```

```python
greeting = "Hello"
name = input("Enter your name: ")
message = input("welcome to Python programming")
result= greeting + name + message
print(result)
```

```python
##String slicing & Indexing

text = input("Enter a string: ")
print(text[0])
print(text[-1])
print(text[0:5])
print(text[-11:])
print(text[::-1])
print(text[11:17])
```

```python
##String Methods
#a. uppercase
sentence = "Python beginner tutorial"
print(sentence.upper())
```

```python
#b. lowercase
sentence = "Python beginner tutorial"
print(sentence.lower())
```

```python
#c. capitalize
sentence = "Python beginner tutorial"
print(sentence.capitalize())
```

```python
#d. count
sentence = "Python beginner tutorial"
print(sentence.count('t'))
```

```python
#e. replace
sentence = "Python beginner tutorial"
print(sentence.replace("Python", "Machine Programming"))
```

```python
##Tuples
t1 = (10, 20, 30)
t2 = (40, 50, 60)

#a. concatenate 2 tuples
combine = t1 + t2
print(combine)
```

```python
t1 = (10, 20, 30)
t2 = (40, 50, 60)

#b. repeat
combine = t1 + t2
print(combine*3)
```

```python
#c. access third element
combine = t1 + t2
print(combine[2])
```

```python
#d. acess first three
combine = t1 + t2
print(combine[:3])
```

```python
#e. access last three
combine = t1 + t2
print(combine[-3:])
```
